```python
import pandas as pd

df = pd.read_csv(
    '/content/drive/MyDrive/Kiva Assignment/kiva_loans.csv'
)

df.head()
```

```python
countries = ['Philippines', 'Kenya', 'El Salvador',
             'Cambodia', 'Pakistan']

selected = df[df['country'].isin(countries)]

selected.groupby('country')['loan_amount'].mean().sort_values(
    ascending=False
)
```

Question 1: Which of the five selected countries has the highest average loan amount?

Interpretation
The analysis shows that El Salvador recorded the highest average loan amount among the five selected countries, with an average of approximately $667.18. This suggests that the average loan size in El Salvador was higher than in the other four countries examined.

```python
df['sector'].value_counts()
```

```python
health_loans = df[
    df['sector'].str.contains('Health', case=False, na=False)
]

health_loans.head()
```

```python
health_above_500 = health_loans[
    health_loans['loan_amount'] > 500
]

print("Total Health loans:", len(health_loans))

print("Health loans above $500:", len(health_above_500))
```

```python
percentage = (
    len(health_above_500) / len(health_loans)
) * 100

print("Percentage above $500:", percentage)
```

Question 2: Are Health-sector loans above $500 common in the Kiva dataset?

Result
Total Health-sector loans 9,223
Loans above $500 5,900
Percentage above $500 63.97%

Interpretation
The analysis shows that 9,223 loans were recorded in the Health sector, of which 5,900 were above $500. This represents approximately 63.97% of all Health-sector loans examined. Therefore, loans exceeding $500 were relatively common in the Health sector within this dataset.
